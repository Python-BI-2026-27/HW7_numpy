#### **Задание 23** (★☆) 🌐

Дана система алгебраических уравнений:

$$ 4\cdot x -2\cdot y = 22 $$
$$ 6\cdot x +  y = 45 $$

Эту СЛАУ можно запиcать в матричном виде:

$$
A\cdot X = B
$$

$$
\begin{pmatrix} 4 & -2 \\ 6 & 1 \end{pmatrix} \cdot X = \begin{pmatrix} 22 \\ 45 \end{pmatrix}
$$

Найдите её решение ($X$) с помощью `np.linalg`


```python
A = np.array([[4, -2], [6, 1]])
B = np.array([22, 45])

result = np.linalg.solve(A, B)
```

```python
assert np.allclose(result, np.array([7., 3.]))
```
